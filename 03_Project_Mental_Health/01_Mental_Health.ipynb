** Checkking Outliers **

```python
numeriical_features = df.select_dtypes(include=[np.number])
numeriical_features
```

```python
# Q1, Q3, IQR
q1 = numeriical_features.quantile(0.25)
q3 = numeriical_features.quantile(0.75)
iqr = q3 - q1
```

```python
Lower_bound = q1 - 1.5 * iqr
Upper_bound = q3 + 1.5 * iqr
```

```python
outliers = (numeriical_features < Lower_bound) | (numeriical_features > Upper_bound)
print(outliers.sum())
```
