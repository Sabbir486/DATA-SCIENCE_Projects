# Project Idea: Sentence Complition (Predict the Next Word)

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import string
```

```python
df = pd.read_csv('qoute_dataset.csv')
df.head()
```

```python
df.shape
```

```python
quotes = df['quote']    # Seperating the quotes column from the dataset and storing it in a variable called
quotes
```

```python
quotes = quotes.str.lower()    # Converting all the quotes to lower case
quotes = quotes.str.translate(str.maketrans('', '', string.punctuation))    # Removing all the punctuation from the quotes
```

```python
quotes
```

# Doing Tokenization

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
vocub_size = 9000    # Setting the vocabulary size to 10000

tokenizer = Tokenizer(num_words=vocub_size, oov_token="<OOV>")
tokenizer.fit_on_texts(quotes)    # Fitting the tokenizer on the quotes
```

```python
word_index = tokenizer.word_index    # Getting the word index from the tokenizer
print(len(word_index))    # Printing the length of the word index
list(word_index.items())[:10]    # Printing the first 10 items of the word index
```

```python
sequences = tokenizer.texts_to_sequences(quotes)    # Converting the quotes to sequences
print(sequences[0])    # Printing the sequences
```

```python
X = []
y = []

for seq in sequences:
    for i in range(1, len(seq)):
        input_seq = seq[:i]    # Getting the sequence from the start to the current index
        output_seq = seq[i]    # Getting the current index
        X.append(input_seq)    # Appending the sequence from the start to the current index to x
        y.append(output_seq)    # Appending the current index to y
```

```python
X
```

```python
y
```

```python
len(X), len(y)    # Printing the length of x and y
```

# Apply Padding on X --> All sentences size will be the same like Largest One

```python
max_len = max([len(x) for x in X])    # Getting the maximum length of the sequences
print(max_len)    
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
X_padded = pad_sequences(X, maxlen=max_len, padding='pre')    # Padding the sequences to the maximum length
print('Before padding:', X[0])
print('After padding:', X_padded[0])    # Printing the first sequence before and after padding
```

```python
y = np.array(y)    # Converting y to a numpy array
```

```python
X_padded.shape
```

# Doing One-Hot Encoding on y

```python
from tensorflow.keras.utils import to_categorical
```

```python
# y_one_hot = to_categorical(y, num_classes=vocub_size)    # There will be one 1 in the index of the word and all other indexes will be 0 in each y column.
y_one_hot = np.asarray(y, dtype=np.int32)
```

```python
y.shape
```

```python
y_one_hot.shape
```

# Creating RNN Model

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, SimpleRNN, GRU
```

```python
embedding_dim = 50    # Setting the embedding dimension to 50
rnn_units = 128    # Setting the number of RNN units (Hidden layers) to 128
```

```python
rnn_model = Sequential()

rnn_model.add(Embedding(input_dim=vocub_size, output_dim=embedding_dim, input_length=max_len))    # Adding an embedding layer to the model

rnn_model.add(SimpleRNN(units=rnn_units, return_sequences=False))    # Adding a Simple RNN layer to the model
rnn_model.add(Dense(units=vocub_size, activation='softmax'))    # Adding a dense layer to the model --> Provides the output of the model

rnn_model.compile(loss='sparse_categorical_crossentropy', optimizer='adam', metrics=['accuracy'])    # Compiling the model
```

```python
rnn_model.summary()
```

# Creating LSTM Model

```python
lstm_model = Sequential()
lstm_units = 128    # Setting the number of LSTM units (Hidden layers) to 128

lstm_model.add(Embedding(input_dim=vocub_size, output_dim=embedding_dim, input_length=max_len))    # Adding an embedding layer to the model
lstm_model.add(LSTM(units=lstm_units, return_sequences=False))    # Adding an LSTM layer to the model
lstm_model.add(Dense(units=vocub_size, activation='softmax'))    # Adding a dense layer to the model --> Provides the output of the model

lstm_model.compile(loss='sparse_categorical_crossentropy', optimizer='adam', metrics=['accuracy'])    # Compiling the model
```

```python
lstm_model.summary()
```

# Implementation Of RNN Model

```python
epochs = 10   
batch_size = 128
```

```python
history_rnn = rnn_model.fit(X_padded, y_one_hot, epochs=epochs, batch_size=batch_size, validation_split=0.2)    # Training the RNN model
```

```python
history_lstm = lstm_model.fit(X_padded, y_one_hot, epochs=epochs, batch_size=batch_size, validation_split=0.2) # Training the LSTM Model
```

```python
index_to_word = {}
for word, index in word_index.items():
    index_to_word[index] = word
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
def predictor(model, tokenizer, text, max_len):
    text = text.lower()

    seq = tokenizer.texts_to_sequences([text])[0]
    seq = pad_sequences([seq], maxlen=max_len, padding='pre')

    pred = model.predict(seq, verbose=0)
    pred_index = np.argmax(pred)
    return index_to_word[pred_index]
```
